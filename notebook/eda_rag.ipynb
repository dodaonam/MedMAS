### Phân phối số từ của tất cả các bài viết

```python
import re
import matplotlib.pyplot as plt

word_count = (
    df["raw_text"]
    .fillna("")
    .apply(lambda x: len(re.findall(r"\S+", x)))
)

plt.figure(figsize=(10, 5))
plt.hist(word_count, bins=60, color="#F58518", edgecolor="white")
plt.title("Distribution of raw_text Length (Words)")
plt.xlabel("Number of Words")
plt.ylabel("Number of Articles")
plt.tight_layout()
plt.show()

word_count.describe(percentiles=[0.5, 0.9, 0.95, 0.99])
```

### Các bài viết có nội dung ít hơn 100 từ

```python
short_mask = word_count < 100
short_articles = (
    df.loc[short_mask, ["article_id", "url", "title", "raw_text"]]
    .assign(word_count=word_count[short_mask].to_numpy())
)

# Dữ liệu plot: gộp theo title để chart gọn hơn
plot_df = (
    short_articles[["title", "word_count"]]
    .assign(title=lambda x: x["title"].fillna("(no title)"))
    .groupby("title", as_index=False)["word_count"]
    .min()
)

# Sắp xếp tăng dần
import numpy as np
order = np.argsort(plot_df["word_count"].to_numpy())
plot_df = plot_df.iloc[order]
```

```python
plt.figure(figsize=(12, max(6, len(plot_df) * 0.35)))
plt.barh(
    plot_df["title"].astype(str).tolist(),
    plot_df["word_count"].to_numpy(),
    color="#E45756"
)
plt.title(f"Articles with raw_text < 100 words")
plt.xlabel("Word Count")
plt.ylabel("Title")
plt.tight_layout()
plt.show()
```

```python
df_qna_100_1000_clean = df_qna_100_1000.copy(deep=True)
print("Original:", len(df_qna_100_1000))
print("Working copy:", len(df_qna_100_1000_clean))
```

```python
# Clean QnA từ df_qna_100_1000_clean -> tạo cặp (anchor, positive)

import re

work = df_qna_100_1000_clean.copy()


def norm_text(s: str) -> str:
    s = str(s).replace("\r\n", "\n").replace("\r", "\n")
    s = re.sub(r"[ \t]+", " ", s)
    s = re.sub(r"\n{3,}", "\n\n", s)
    return s.strip()


def find_marker_line(lines, marker: str, start: int = 0):
    """
    Tìm marker khi marker đứng riêng 1 dòng.
    Không dùng regex để tránh bắt nhầm chữ 'hỏi' trong 'câu hỏi', 'cho em hỏi', ...
    """
    marker = marker.casefold()

    for i in range(start, len(lines)):
        if lines[i].strip().casefold() == marker:
            return i

    return None


def is_identity_line(x: str) -> bool:
    x = x.strip()

    if not x:
        return True

    # Không xóa dòng có vẻ là câu hỏi thật
    if "?" in x or re.search(r"(?i)\b(cho em hỏi|xin hỏi|hỏi bác sĩ)\b", x):
        return False

    # Dòng định danh / credit chắc chắn nên bỏ
    identity_patterns = [
        r"(?i)^khách hàng ẩn danh$",
        r"(?i)^câu hỏi ẩn danh$",
        r"(?i)^được giải đáp bởi\b.*$",
        r"(?i)^được tư vấn bởi\b.*$",
        r"(?i)^bài viết được tư vấn bởi\b.*$",
        r"(?i)^ths\..*$",
        r"(?i)^bs\..*$",
        r"(?i)^bsck\b.*$",
        r"(?i)^thạc sĩ\b.*$",
        r"(?i)^bác sĩ chuyên khoa\b.*$",
        r"(?i)^phó giáo sư\b.*$",
        r"(?i)^tiến sĩ\b.*$",
    ]

    if any(re.match(p, x) for p in identity_patterns):
        return True

    # Dòng tên + năm sinh / địa danh
    if re.fullmatch(r".{1,80}\(\s*(\d{4}|[^)]{2,30})\s*\)", x):
        return True

    return False


def drop_identity_lines(lines):
    return [
        x.strip()
        for x in lines
        if x.strip() and not is_identity_line(x)
    ]


def cut_tail_boilerplate(answer: str) -> str:
    stops = [
        r"(?im)^[ \t]*nếu bạn còn thắc mắc\b",
        r"(?im)^[ \t]*cảm ơn .*vinmec",
        r"(?im)^[ \t]*trân trọng!?\.?[ \t]*$",
        r"(?im)^[ \t]*các nội dung tư vấn khác\b",
        r"(?im)^[ \t]*các bài viết cùng chủ đề\b",
        r"(?im)^[ \t]*kết luận[ \t]*:[ \t]*$",
    ]

    cut = len(answer)

    for p in stops:
        m = re.search(p, answer)
        if m:
            cut = min(cut, m.start())

    answer = answer[:cut].strip()

    # Bỏ tag/category cuối bài
    lines = [x.strip() for x in answer.split("\n") if x.strip()]

    while lines:
        last = lines[-1]

        if last.casefold() == "qna":
            lines.pop()
            continue

        # Xóa các dòng tag ngắn cuối bài, ví dụ: Nhi, Vacxin, Đau lưng...
        # Nhưng giữ lại câu thật nếu có dấu câu.
        if len(last) <= 45 and not re.search(r"[.!?:,]", last):
            lines.pop()
            continue

        break

    return "\n".join(lines).strip()


def extract_qna(raw_text: str):
    t = norm_text(raw_text)
    lines = t.split("\n")

    q_idx = find_marker_line(lines, "Hỏi")
    if q_idx is None:
        return "", "", False, "missing_question_marker"

    a_idx = find_marker_line(lines, "Trả lời", start=q_idx + 1)
    if a_idx is None:
        return "", "", False, "missing_answer_marker"

    if a_idx <= q_idx:
        return "", "", False, "marker_order_invalid"

    q_raw = "\n".join(lines[q_idx + 1:a_idx]).strip()
    a_raw = "\n".join(lines[a_idx + 1:]).strip()

    q_lines = drop_identity_lines(q_raw.split("\n"))
    a_lines = drop_identity_lines(a_raw.split("\n"))

    q_clean = norm_text("\n".join(q_lines))
    a_clean = norm_text(cut_tail_boilerplate("\n".join(a_lines)))

    if len(q_clean) < 10:
        return q_clean, a_clean, False, "anchor_too_short"

    if len(a_clean) < 25:
        return q_clean, a_clean, False, "positive_too_short"

    return q_clean, a_clean, True, "ok"


parsed = work["raw_text"].apply(extract_qna)

work["anchor"] = parsed.apply(lambda x: x[0])
work["positive"] = parsed.apply(lambda x: x[1])
work["qa_parse_ok"] = parsed.apply(lambda x: x[2])
work["fail_reason"] = parsed.apply(lambda x: x[3])

df_qna_pairs_clean = work.loc[
    work["qa_parse_ok"],
    ["article_id", "url", "title", "anchor", "positive"]
].copy()

print("Input:", len(work))
print("Parse OK:", len(df_qna_pairs_clean))

print("\nFail reasons:")
display(work.loc[~work["qa_parse_ok"], "fail_reason"].value_counts())

display(df_qna_pairs_clean.head(10))
```

```python
# 1) In bài fail
failed_rows = work.loc[~work["qa_parse_ok"], ["article_id", "url", "title", "fail_reason", "raw_text", "anchor", "positive"]].copy()
print("Failed rows:", len(failed_rows))
display(failed_rows.head(5))
```

```python
# 2) Final clean cho embedding + khử duplicate exact theo cặp (anchor, positive)

import re

def clean_for_embedding(s: str) -> str:
    s = str(s).replace("\r\n", "\n").replace("\r", "\n")
    s = re.sub(r"\s+", " ", s)   # gộp toàn bộ space, tab, newline thành 1 space
    return s.strip()


before = len(df_qna_pairs_clean)

df_embedding_pairs = df_qna_pairs_clean[["anchor", "positive"]].copy()

# Clean text cuối cùng để đưa vào embedding model
df_embedding_pairs["anchor"] = df_embedding_pairs["anchor"].apply(clean_for_embedding)
df_embedding_pairs["positive"] = df_embedding_pairs["positive"].apply(clean_for_embedding)

# Bỏ dòng rỗng nếu có
df_embedding_pairs = df_embedding_pairs[
    (df_embedding_pairs["anchor"].str.len() > 0) &
    (df_embedding_pairs["positive"].str.len() > 0)
].copy()

# Khử duplicate exact sau khi đã normalize text
df_embedding_pairs = (
    df_embedding_pairs
    .drop_duplicates(subset=["anchor", "positive"])
    .reset_index(drop=True)
)

after = len(df_embedding_pairs)

print("Before final clean/dedup:", before)
print("After final clean/dedup :", after)
print("Removed rows           :", before - after)

display(df_embedding_pairs.head())
```

```python
# 3) Lưu file
from pathlib import Path
import json

out_path = Path("../data/proccessed/vinmec_qna_embedding_pairs.jsonl")
out_path.parent.mkdir(parents=True, exist_ok=True)

with out_path.open("w", encoding="utf-8") as f:
    for record in df_embedding_pairs.to_dict(orient="records"):
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

print("Saved:", out_path.resolve())
print("Rows :", len(df_embedding_pairs))
```
